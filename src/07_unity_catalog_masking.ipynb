```python
%sql
USE CATALOG logistics_prod;

-- 1. Creamos una función de enmascaramiento dinámico para datos financieros
CREATE OR REPLACE FUNCTION logistics_prod.gold.mask_financial_value(amount DOUBLE)
RETURN CASE
    WHEN is_account_group_member('finance_auditors') THEN amount
    ELSE NULL -- Oculta el valor exacto mostrando NULL a usuarios sin permiso financiero
END;

-- 2. Aplicamos la máscara directamente sobre la columna de la tabla Gold
ALTER TABLE logistics_prod.gold.asset_performance_summary
ALTER COLUMN total_cargo_value SET MASK logistics_prod.gold.mask_financial_value;
```
